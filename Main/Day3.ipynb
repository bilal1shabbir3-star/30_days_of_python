```python
# Getting the len of the set

len(fruits)
```

```python
# Checking an item

fruits = {'banana', 'orange', 'mango', 'lemon'}

print('banana' in fruits)
```

```python
# Adding items to the set

fruits = {'banana', 'orange', 'mango', 'lemon'}

fruits.add('apple')
fruits
```

```python
# To add multiple items in the set use the update() function

fruits = {'banana', 'orange', 'mango', 'lemon'}

fruits.update(['apple','lime'])
fruits
```

```python
# Removing items from a set

fruits = {'banana', 'orange', 'mango', 'lemon'}

fruits.remove('banana') # you can also use the pop function for this purpose
fruits
```

```python
# Clearing items in a set

fruits = {'banana', 'orange', 'mango', 'lemon'}

fruits.clear()

fruits
```

```python
# Deleting a set

fruits = {'banana', 'orange', 'mango', 'lemon'}

del fruits
```

```python
# Converting list to a set

fruits_lst = ['banana', 'orange', 'mango', 'lemon']

fruits_st = set(fruits_lst)

type(fruits_lst), type(fruits_st)
```

```python
# Joining sets

fruits = {'banana', 'orange', 'mango', 'lemon'}
vegetables = {'tomato', 'potato', 'chili', 'garlic'}

combo_st = fruits.union(vegetables)

combo_st
```

```python
fruits.update(vegetables)
fruits
```
