# List

```python
fruits = ['banana', 'orange', 'mango', 'lemon','lime','apple']
countries = ['Germany', 'France','Belgium','Sweden','Denmark','Finland','Norway','Iceland','Estonia']
```

```python
# Unpacking the list

first_fruit,second_fruit,third_fruit,*rest = fruits

print('fruits',fruits)
print(f'\nFirst Fruit: {first_fruit}\n')
print(f'Second Fruit: {second_fruit}\n')
print(f'Third Fruit: {third_fruit}\n')
print(f'Rest of the Fruits: {rest}\n')
```

```python
# Slicing items from the lists

# Positive indexing

fruits

all_fruits1 = fruits[0:] # from the 0th index till the last index of the list
all_fruits2 = fruits[0:len(fruits)] # same result as above one
all_fruits3 = fruits[0:6] # same as above

print(all_fruits1)
print(all_fruits2)
print(all_fruits3)

all_fruits1 == all_fruits2 == all_fruits3
```

```python
# Slicing items from the lists

# Negative indexing

fruits

all_fruits1 = fruits[-6:] # from the 0th index till the last index of the list
all_fruits2 = fruits[-len(fruits):] # same result as above one

print(all_fruits1)
print(all_fruits2)

all_fruits1 == all_fruits2 
```

```python
# Modifying Lists

fruits = ['banana', 'orange', 'mango', 'lemon']

fruits[0] = 'apple'

fruits
```

```python
# Checking an item in the list

fruits = ['banana', 'orange', 'mango', 'lemon']

does_exist = 'orange'  in fruits

does_exist
```

```python
# Adding an item in the list

fruits = ['banana', 'orange', 'mango', 'lemon']

fruits.append('apple')

fruits
```

```python
# Inserting an item in the list

fruits = ['banana', 'orange', 'mango', 'lemon']

fruits.insert(1,'apple') # inserting a single item

print("Single item insertion" , fruits)
```

```python
# Removing an item from the list

fruits = ['banana', 'orange', 'mango', 'lemon']

fruits.remove('banana')

print(fruits)
```

```python
# We can also remove the item using the pop function

fruits = ['banana', 'orange', 'mango', 'lemon']

fruits.pop(0) # here we can pass index not the string or item we want to remove

print(fruits)
```

```python
# Del can also be use to delete an item

fruits = ['banana', 'orange', 'mango', 'lemon']

fruits.__delitem__(0)

print(fruits)

del fruits[2]

print(fruits)
```

```python
# We use clear to clear all of the list

fruits = ['banana', 'orange', 'mango', 'lemon']

fruits.clear()

fruits
```

```python
# Copying a list

"""

If we assign a list to the other variable than any changes in that list of second variable will also gonna change the original list.
There will be cases when we don't want to make any changes in the original list but in the copy one, to do that we will use the copy()
function to make copies of list.

"""

fruits = ['banana', 'orange', 'mango', 'lemon']

copy_of_fruits = fruits.copy()

copy_of_fruits.append('apple')

print("Original List: " , fruits)
print("\nCopy List: " , copy_of_fruits)
```

```python
# Joining lists

"""
Syntax:

        1) using the plus(+) operator
        
        list = lst1 + lst2
        
        2) using the extend function
        
        lst1.extend(lst2)

"""

fruits = ['banana', 'orange', 'mango', 'lemon','lime','apple']
countries = ['Germany', 'France','Belgium','Sweden']

combined_using_plus = fruits + countries # storing in a new vairable

print(combined_using_plus)

countries.extend(fruits) # extend in the countries list without creating the second list

print(countries)
```

```python
# counting the items

fruits = ['banana', 'orange', 'mango', 'lemon','lime','apple']

print(f'Count: {fruits.count('orange')}')


# reversing the list

fruits.reverse()

print(f'Reversed list: {fruits}')


# finding index of an item

print(f'Index of an item in the list: {fruits.index('orange')}')
```
