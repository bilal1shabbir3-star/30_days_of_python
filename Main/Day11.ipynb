```python
# Python has datetime module to handle date and time
import datetime
print(dir(datetime))
```

```python
# Getting datetime information

from datetime import datetime

now = datetime.now()
print(now)

day = now.day
print(day)

month = now.month
print(month)

year = now.year
print(year)

hour = now.hour
print(hour)

minutes = now.min
print(minutes)

seconds = now.second
print(seconds)

timestamp = now.timestamp()
print(timestamp)

print(day, month, year, hour, minutes, seconds)
print('timestamp', timestamp)
print(f'{day}/{month}/{year}, {hour}:{minutes}')
```

```python
# Continue

from datetime import datetime

# Current  date and time

now = datetime.now()
t = now.strftime("%H:%M:%S")
print('time:', t)

time_one = now.strftime("%m/%d/%Y, %H:%M:%S")
print("TimeOne:", time_one)
```

```python
# String to time using strptime

from datetime import datetime

date_string = "24 September, 2026"
print("datetime_string =", date_string)

date_obj = datetime.strptime(date_string, "%d %B, %Y")
print("date_obj =",date_obj)
```

```python
# Using date from datetime

from datetime import date

d = date(2005, 1, 1)
print("Date =", d)

today = d.today()
print("Current_date =", today)

print("Current_year =", today.year)
print("Current_month =", today.month)
print("Current_day =", today.day)
```

```python
# Time objects to represent

from datetime import time

a = time()
print("a:", a)

b = time(6, 42, 32)
print("b:", b)

c = time(hour= 6, minute= 42, second= 32)
print("c:", c)

d =time(6, 42, 32, 132354)
print("d:", d)
```

```python
# Difference between two points in Time

from datetime import date, datetime

today = date(2026, 9, 25)
print(today)
new_year = date(year=2027, month=1, day=1)
time_left_for_new_year = new_year - today

print("Time left for new year is :", time_left_for_new_year)

t1 = datetime(year=2026, month=9, day=25, hour=17, minute=58, second=0)
t2 = datetime(year=2027, month=1, day=1, hour=0, minute=0, second=0)
diff = t2 - t1
print("Time left for new year is :", diff)
```

```python
# Difference Between two points in Time using timedelta

from datetime import timedelta

t1 = timedelta(weeks=12, days=10, hours=4, seconds=20)
t2 = timedelta(days=7, hours=5, minutes=3, seconds=30)
t3 = t1 - t2
print("t3:", t3)
```
