# Tuples

```python
# creating an empty tuple

empty_tuple = ()

# we can also use the tuple() function for this as well

mty_tuple = tuple()

print(empty_tuple)
print(mty_tuple)
```

```python
# tuple with initial values

tpl = ('meow','hello','world')

print(tpl)

# length of the tuple

print("Length of the tuple: " , len(tpl))
```

```python
# Accessing the tuple items

fruits = ('banana', 'orange', 'mango', 'lemon')

fruits[1]
```

```python
# negative indexing to access the items in the tuple

fruits = ('banana', 'orange', 'mango', 'lemon')

fruits[-1] # for the last item in the tuple
```

```python
# Slicing the tuples

fruits = ('banana', 'orange', 'mango', 'lemon')

all_fruits = fruits[0:] # from the 0th index till the last index in the tuple

banana_orange = fruits[0:2] # the result will be from the 0th index to the index before 2, 2nd index item will not appear in the result

orange_mango = fruits[1:3]

print("All fruits: " , all_fruits)
print("\n Banana to Orange: " , banana_orange)
print("\n Orange to mango f: " , orange_mango)
```

```python
# Negative index slicing 

fruits = ('banana', 'orange', 'mango', 'lemon')

all_fruits = fruits[:] # from the last index till the first index in the tuple

banana_orange = fruits[-4:-1]

orange_mango = fruits[-3:-1]

print("All fruits: " , all_fruits)
print("\n Banana to Orange: " , banana_orange)
print("\n Orange to mango f: " , orange_mango)
```

```python
# Changing tuples to list

fruits = ('banana', 'orange', 'mango', 'lemon')

print("Data type before: " , type(fruits))

fruits = list(fruits)

print("Data type after: " , type(fruits))
```

```python
# checking an item in the list

fruits = ('banana', 'orange', 'mango', 'lemon')

does_exist = 'banana' in fruits

print(does_exist)
```

```python
# joining tuples

fruits = ('banana', 'orange', 'mango', 'lemon','lime','apple')
countries = ('Germany', 'France','Belgium','Sweden')

combined_tpl = fruits + countries

print("Combination of two tuples: " , combined_tpl)
```

```python
# deleting tuples

fruits = ('banana', 'orange', 'mango', 'lemon','lime','apple')

del fruits

print(fruits) # will result an error as we are deleting the tuple in the above line
```
