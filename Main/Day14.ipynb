# File Handling:

So far we have seen different Python data types. We usually store our data in different file formats. In addition to handling files, we will also see different file formats(.txt, .json, .xml, .csv, .tsv, .excel) in this section. First, let us get familiar with handling files with common file format(.txt).

File handling is an import part of programming which allows us to create, read, update and delete files. In Python to handle data we use open() built-in function.

**Syntax**

open('filename', mode) # mode(r, a, w, x, t,b)  could be to read, write, update

* "r" - Read - Default value. Opens a file for reading, it returns an error if the file does not exist
* "a" - Append - Opens a file for appending, creates the file if it does not exist
* "w" - Write - Opens a file for writing, creates the file if it does not exist
* "x" - Create - Creates the specified file, returns an error if the file exists
* "t" - Text - Default value. Text mode
* "b" - Binary - Binary mode (e.g. images)

```python
'''
--Opening Files for Reading--

The default mode of open is reading, so we do not have to specify 'r' or 'rt'. 
I have created and saved a file named reading_file_example.txt in the files directory.
Let us see how it is done:
'''

f = open('Text_Document.txt')
txt = f.read()
print(type(txt))
print(txt)
f.close
```

```python
# Printing the first 10 characters of the text file

f = open("Text_Document.txt")
txt = f.read(10)
print(txt)
f.close()
```

```python
# "readline()" reads only the first line of the text file

f = open("Text_Document.txt")
txt = f.readline()
print(txt)
f.close()
```

```python
# "readlines()" reads all the text line by line and returns a list of lines
 
f = open("Text_Document.txt")
txt = f.readlines()
print(txt)
f.close()
```

```python
# Another way to get all the lines as a list is using splitlines():

f = open("Text_Document.txt")
txt = f.read().splitlines()
print(txt)
f.close()
```

```python
"""
After we open a file, we should close it. There is a high tendency of forgetting to close them.
There is a new way of opening files using with - closes the files by itself.
Let us rewrite the the previous example with the with method:

"""

with open("Text_Document.txt") as f :
    lines = f.read().splitlines()
    print(lines)
```

**Opening Files for Writing and Updating**

To write to an existing file, we must add a mode as parameter to the open() function:

* "a" - append - will append to the end of the file, if the file does not exist it creates a new file.
* "w" - write - will overwrite any existing content, if the file does not exist it creates.

```python
# Let us append some text to the files we have been reading

with open("Text_Document.txt", 'a') as f:
    f.write("This text has to be appended at the end")
```

```python
# The method below will create a new file

with open("Text_Document.txt", 'w') as f:
    f.write("This text will be written in a newly created file")
```

```python
# Deleting Files

"""
We have seen in previous section, how to make and remove a directory using os module.
Again now, if we want to remove a file we use os module.

"""

import os
os.remove("Text_Document copy.txt")
```

```python
# If the file is already removed than the above cell of code will raise an error, so it is good to use a condition like this

import os

if os.path.exists("Text_Document copy.txt"):
    os.remove("Text_Document copy.txt")
else:
    print("The file does not exist")
```

**File with json Extension**

            JSON stands for JavaScript Object Notation. Actually, it is a stringified JavaScript object or Python dictionary.

```python
# dictionary

person_dct = {
    "Name"      : "Bilal Shabbir",
    "Age"       :  21,
    "Country"   :  "Pakistan",
    "City"      :  "Sadiqabad"
}

# JSON: A string form a dictionary

person_json = "{'Name':'Bilal Shabbir','Age':  21,'Country':'Pakistan','City':'Sadiqabad'}"

# We use three quotes to make it more readable

person_json = '''
{
    "Name"      : "Bilal Shabbir",
    "Age"       :  21,
    "Country"   :  "Pakistan",
    "City"      :  "Sadiqabad"
}
'''
```

```python
# Changing JSON to dictionary

import json

person_json = '''
{
    "Name"      : "Bilal Shabbir",
    "Age"       :  21,
    "Country"   :  "Pakistan",
    "City"      :  "Sadiqabad"
}
'''

# lets change JSON to dictionary

person_dct = json.loads(person_json)
print(type(person_dct))
print(person_dct)
print(person_dct['Name'])
```

```python
# Changing dictionary to JSON

import json

# python dictionary

person_dct = {
    "Name"      : "Bilal Shabbir",
    "Age"       :  21,
    "Country"   :  "Pakistan",
    "City"      :  "Sadiqabad"
}

# lets convert it to json

perosn_json = json.dumps(person_dct, indent=4) # indent could be 2, 4, 8. It beautifies the JSON
print(type(person_json))
print(person_json)
```

```python
# Saving as JSON file

'''
We can also save our data as a json file. Let us save it as a json file using the following steps. 
For writing a json file, we use the json.dump() method, it can take dictionary, output file, ensure_ascii and indent.

'''

import json

# python dictionary

person = {
    "Name"      : "Bilal Shabbir",
    "Age"       :  21,
    "Country"   :  "Pakistan",
    "City"      :  "Sadiqabad"
}

with open("Saving_JSON_file.json", 'w') as f:
    json.dump(person, f, ensure_ascii=False, indent=4)
```

In the code above, we use encoding and indentation. Indentation makes the json file easy to read.

**File with csv Extension**

            CSV stands for comma separated values. CSV is a simple file format used to store tabular data, such as a spreadsheet or database. CSV is a very common data format in data science.

**For example**

            "Name","Age","Country","City",
            "Bilal Shabbir",21,"Pakistani","Sadiqabad"

```python
# Lets create a csv file first

import csv

headers = ["Name", "Age", "Country", "City"]

rows = [
    ["Bilal Shabbir", 21, "Pakistan", "Sadiqabad"],
    ["Qosain Shahzad", 21, "Pakistan", "Sadiqabad"],
    ["Jam Murad", 22, "Pakistan", "Sadiqabad"],
    ]

with open("person_info.csv", "w", newline="") as f:
    writer  = csv.writer(f) # writing the csv file
    # Write the header rows
    writer.writerow(headers)
    # Write all data rows at once
    writer.writerows(rows)
    
print("CSV file created successfully!")
```

```python
import csv

with open("person_info.csv") as f:
    csv_reader = csv.reader(f, delimiter= ",") # we use, read method to read csv
    line_count = 0
    for row in csv_reader:
        if line_count == 0:
            print(f'Column names are :{", ".join(row)}')
            line_count +=1
        else:
            print(
                f'\t{row[0]} is a teacher. He lives in {row[2]}, {row[3]}.')
            line_count += 1
    print(f"The number of lines are: {line_count}")
```

### File with xlsx Extension
            To read excel files we need to install xlrd package. We will cover this after we cover package installing using pip.



```python
# Creating an xlsx file

import csv
from openpyxl import Workbook

# Create a new excel book and create a new sheet
wb = Workbook()
ws = wb.active

# open the csv file and append each row from csv file excel
with open("person_info.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        ws.append(row)
        
# Save the workbook as an xlsx file
wb.save("sample.xlsx")
```
