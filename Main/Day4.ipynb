# Dictionaries:

```python
# creating an empty dictionary

empty_dict = {}

empty_dict
```

```python
# creating first dictionary

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

type(person), person
```

```python
# length of dictionary

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

print(len(person))
```

```python
# Accessing Dictionary Items

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

print(person["Name"])
print(person["Age"])
print(person["Gender"])
```

Accessing an item by key name raises an error if the key does not exist. To avoid this error first we have to check if a key exist or we can use the get method. The get method returns None, which is a NoneType object data type, if the key does not exist.

```python
# Accessing Dictionary Items

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

print(person.get("Name"))
print(person.get("Age"))
print(person.get("Gender"))
print(person.get("gender"))
print(person["gender"])
```

```python
# Adding items to a dictionary

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

person['Weight_in_kg'] = 72

person
```

```python
# Modifying items in a dictionary

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

person["Name"] = "Bilal Shabbir"

person
```

```python
# Checking keys in a dictionary

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

print('Name' in person)
print('Bilal' in person) # False, it is not the key
print('Gender' in person)
```

```python
# Removing key and value pair from dictionary

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

del person['Gender']

person
```

```python
# Dictionary to List of items

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

person.items()
```

```python
# Clearing the dictionary

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

person.clear()

person
```

```python
# Deleting the dictionary

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

del person

person
```

```python
# Copying the dictionary

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

person_copy = person.copy()

print("Before")
print("-------------------")
print("Original: ", person)
print("Copy: ", person_copy)

del person_copy['Gender']

print("\nAfter")
print("-------------------")
print("Original after changing the copy: ", person)
print("Copy after doing changes: ", person_copy)
```

```python
# Getting dictionary keys as list

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

print(person.keys())
```

```python
# Getting dictionary values as list

person = {
    
    "Name":"Bilal",
    "Age":21,
    "Gender":"Male",
}

print(person.values())
```
